```python

facts = {
    "barks(Tommy)",
    "has_four_legs(Tommy)",
    "has_sharp_teeth(Tommy)"
}

rules = [
    # Rule 1: If X barks and has four legs, X is a dog
    (["barks(Tommy)", "has_four_legs(Tommy)"], "dog(Tommy)"),

    # Rule 2: If X has feathers and lays eggs, X is a bird
    (["has_feathers(Tommy)", "lays_eggs(Tommy)"], "bird(Tommy)"),

    # Rule 3: If X is a dog, X is a mammal
    (["dog(Tommy)"], "mammal(Tommy)"),

    # Rule 4: If X is a mammal and has sharp teeth, X is a carnivore
    (["mammal(Tommy)", "has_sharp_teeth(Tommy)"], "carnivore(Tommy)"),

    # Rule 5: If X is a bird, X can fly
    (["bird(Tommy)"], "can_fly(Tommy)")
]

def forward_chaining(facts, rules):
    inferred = True

    while inferred:
        inferred = False

        for conditions, conclusion in rules:
            if all(condition in facts for condition in conditions):
                if conclusion not in facts:
                    facts.add(conclusion)
                    print("New fact inferred:", conclusion)
                    inferred = True

    return facts


print("Initial Facts:")
for fact in facts:
    print(fact)

print("\nApplying Forward Chaining...\n")

final_facts = forward_chaining(facts, rules)

print("\nFinal Result:")

if "carnivore(Tommy)" in final_facts:
    print("Tommy is a CARNIVORE.")
else:
    print("Tommy is NOT PROVEN to be a carnivore.")
```
